```python
# Carga del dataset
import pandas as pd

dataframe = pd.read_csv(
    '../data/sentences.csv.zip',
    index_col=False,
    compression='zip'
)
```

```python
# Ejemplos de frases neutrales
for i in range(5):
    print(dataframe[dataframe.target == 'neutral']['phrase'].iloc[i])
```

```python
# Particionamiento del dataset

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    dataframe.phrase,
    dataframe.target,
    test_size=0.3,
    shuffle=False
)
```

```python
# Construcción de la matriz documento-término (DTM) con CountVectorizer

from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(
    lowercase=True,
    analyzer='word',
    token_pattern=r"\b[a-zA-Z]\w+\b",
    stop_words='english',
    max_df=0.99,
    min_df=2,
    binary=True
)

vectorizer.fit(X_train)

vectorizer.get_feature_names_out()
```

```python
# Entrenamiento del modelo de clasificación

from sklearn.linear_model import LogisticRegression

clf = LogisticRegression()

X_train_vectorized = vectorizer.transform(X_train)

clf.fit(X_train_vectorized, y_train)
```

```python
# Evaluación del modelo de clasificación

from sklearn.metrics import accuracy_score

# Muestra de entrenamiento

accuracy_score(
    y_true=y_train,
    y_pred=clf.predict(X_train_vectorized)
)
```

```python
# Mustra de prueba

X_test_vectorized = vectorizer.transform(X_test)
predictions = clf.predict(X_test_vectorized)

accuracy_score(
    y_true=y_test,
    y_pred=predictions
)
```

```python
# Matriz de confusión

from sklearn.metrics import ConfusionMatrixDisplay

display = ConfusionMatrixDisplay.from_predictions(
    y_test,
    predictions,
    cmap='Greens',
)
display.figure_.suptitle('Matriz de confusión')
print(f'Confusion matrix:\n{display.confusion_matrix}')
```

Se usa balanced accuracy cuando entre las categorías las muestras son muy diferentes en tamaño

```python
# Almacenamiento del modelo entrenado y del vectorizador

import pickle

with open('../submission/clf.pkl', 'wb') as f:
    pickle.dump(clf, f)

with open('../submission/vectorizer.pkl', 'wb') as f:
    pickle.dump(vectorizer, f)
```

```python
# Abrir modelo

with open('../submission/clf.pkl', 'rb') as f:
    new_clf = pickle.load(f)

with open('../submission/vectorizer.pkl', 'rb') as f:
    new_vectorizer = pickle.load(f)

accuracy_score(
    y_true=dataframe.target,
    y_pred=new_clf.predict(new_vectorizer.transform(dataframe.phrase))
)
```
